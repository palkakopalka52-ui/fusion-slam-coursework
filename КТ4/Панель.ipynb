# КТ4 — выбор шага
Камера пропускает середину маршрута. Меняю seed и правило выбора демпфирования в общей библиотеке.

```python
%matplotlib inline
import numpy as np, matplotlib.pyplot as plt, ipywidgets as w
from study_neuro import episode, solve, score, DampingAdvisor
from IPython.display import display

def compare(seed, method):
    graph, reference = episode(seed, 'dropout')
    advisor = DampingAdvisor.load('results/'+method+'.npz') if method != 'classical' else method
    est = solve(graph, policy=advisor, initial_lambda=.0003, budget=24)
    fig, axes = plt.subplots(1,2,figsize=(8,2.4),layout='constrained')
    for path, label in [(reference.poses,'Эталон'), (est.poses,'Оценка')]:
        axes[0].plot(*path[:,:2].T,label=label)
    axes[0].set(xlabel='x, м',ylabel='y, м',aspect='equal'); axes[0].legend()
    axes[1].plot(est.history,'o-',color='#007879'); axes[1].set(xlabel='Принятые шаги',ylabel='Значение цели')
    plt.show()
    print(f'ATE {score(est.poses,reference)["ate_m"]:.4f} м, решений {est.solves}, отказов {est.rejects}')

w.interact(compare,seed=w.IntSlider(value=1810,min=1810,max=1817,description='Маршрут'),
           method=w.Dropdown(options=['classical','mlp','rnn'],value='rnn',description='Метод'));
```
